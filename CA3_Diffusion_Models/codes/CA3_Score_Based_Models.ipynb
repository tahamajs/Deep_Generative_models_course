# CA3: Score-based Generative Models Implementation

## 📋 Overview

This notebook implements **Score-based Generative Models** using Langevin dynamics for sampling from complex distributions. The implementation focuses on 2D Gaussian mixture data, demonstrating score matching, noise-conditioned score prediction, and various sampling strategies.

Score-based generative models learn the gradient of the log-probability density (score function) rather than the density itself, enabling efficient sampling via Langevin dynamics. This approach avoids the normalization constant problem that plagues density estimation.

---

## 🎯 Main Concepts at a Glance

### 1. **Score Functions** 
The score function $s(x) = \nabla_x \log p(x)$ represents the gradient of log-probability. Instead of learning the entire probability density (which requires expensive normalization), we learn just the gradient - which is much easier!

**Key Advantage**: No normalization constant needed! We can sample directly by following gradients.

### 2. **Denoising Score Matching (DSM)**
We train the network to predict scores of **noisy** data. The true score of a noisy distribution $p_\sigma(x | x_0) = \mathcal{N}(x | x_0, \sigma^2 I)$ is simply $-\frac{(x - x_0)}{\sigma^2}$, which points back to the original clean data.

**Training Objective**: 
$$\min_\theta \mathbb{E}\left[\left\|s_\theta(x+\epsilon, \sigma) + \frac{\epsilon}{\sigma^2}\right\|^2\right]$$

### 3. **Noise-Conditioned Score Networks**
The network takes both **data** and **noise level** (σ) as input: $s_\theta(x, \sigma)$. This allows:
- Training on multiple noise scales simultaneously
- Flexibility to sample at different noise levels
- Robustness across noise variations

### 4. **Langevin Dynamics Sampling**
Once we have a learned score function, we can generate samples by following the gradient:
$$x_{t+1} = x_t + \frac{\epsilon}{2}s_\theta(x_t, \sigma) + \sqrt{\epsilon} z_t$$

This is like hiking uphill toward probability peaks, with some random exploration.

### 5. **Annealed Sampling**
Start with high noise (easy exploration) and gradually reduce noise (precise refinement):
- **High σ**: Smooth distribution, easy to explore
- **Low σ**: Sharp distribution, precise sampling
- **Annealing**: Best of both worlds!

**Denoising Score Matching Objective**:
$$\min_\theta \mathbb{E}_{x \sim p_{data}, \epsilon \sim \mathcal{N}(0,σ^2I)} \left[\left\|s_\theta(x+\epsilon, \sigma) + \frac{\epsilon}{\sigma^2}\right\|^2\right]$$

**Langevin Dynamics**:
$$x_{t+1} = x_t + \epsilon \nabla_x \log p(x_t) + \sqrt{2\epsilon} z_t, \quad z_t \sim \mathcal{N}(0, I)$$

### Function: `generate_mixture_gaussians()`

#### Parameters
- **n_samples** (default: 5000): Total number of samples to generate

#### What It Does

1. **Component Weight Generation**:
   - Randomly generates weight w₁ ∈ [0.15, 0.25]
   - w₂ = 1 - w₁
   - Randomly swaps weights between components

2. **Mean Selection**:
   - Randomly chooses between two configurations:
     - Configuration 1: μ₁ = [-5, 5], μ₂ = [5, -5]
     - Configuration 2: μ₁ = [5, 5], μ₂ = [-5, -5]

3. **Covariance Matrix**:
   - Each component has diagonal covariance: diag(1 + random, 1 + random)
   - Ensures non-zero variance for both dimensions

4. **Sample Generation**:
   - Samples n₁ points from first Gaussian (n₁ = w₁ × n_samples)
   - Samples n₂ points from second Gaussian (n₂ = w₂ × n_samples)
   - Concatenates and randomly permutes

5. **Train/Test Split**:
   - Randomly splits data according to train_split ratio

#### Returns
- **train_samples**: Training data points (2D numpy array)
- **test_samples**: Test data points (2D numpy array)

## 🔬 Denoising Score Matching: Step-by-Step Demonstration

### The Core Idea

**Denoising Score Matching (DSM)** trains the network to predict the score of noisy data:

$$\text{Loss} = \frac{1}{2}\mathbb{E}\left[\left\|s_\theta(x+\epsilon, \sigma) + \frac{\epsilon}{\sigma^2}\right\|^2\right]$$

Where:
- $x$ = clean data
- $\epsilon \sim \mathcal{N}(0, \sigma^2 I)$ = noise
- $x + \epsilon$ = noisy data
- $s_\theta(x+\epsilon, \sigma)$ = predicted score
- $-\epsilon/\sigma^2$ = **true score** of noisy distribution

### Why This Works

The true score of $p_\sigma(x | x_0) = \mathcal{N}(x | x_0, \sigma^2 I)$ is:
$$\nabla_x \log p_\sigma(x | x_0) = -\frac{(x - x_0)}{\sigma^2} = -\frac{\epsilon}{\sigma^2}$$

By predicting this, we indirectly learn to point back to the original (clean) data point $x_0$!

## 🧠 Score Network Architecture (ScoreNet)

### Overview

This cell defines the **ScoreNet** architecture, a neural network that takes noisy data and noise level as input and outputs the predicted score function (gradient of log-density). The architecture is a multi-layer perceptron (MLP) with batch normalization and LeakyReLU activations.

---

### Architecture Details

#### Network Structure

```
Input: [x, σ] (concatenated) → 3D vector
    ↓
Linear(3 → 256) + BatchNorm + LeakyReLU
    ↓
[Repeat 4 times:]
    Linear(256 → 256) + BatchNorm + LeakyReLU
    ↓
Linear(256 → 2)
    ↓
Output: Score vector (2D)
```

#### Layer Specifications

- **Input Dimension**: 3 (2D data + 1D noise level)
- **Hidden Dimension**: 256 neurons per layer
- **Number of Hidden Layers**: 4 (6 total layers including input/output)
- **Output Dimension**: 2 (2D score vector)

#### Key Design Choices

##### 1. **Input Concatenation**
- Data $x \in \mathbb{R}^2$ and noise level $\sigma \in \mathbb{R}$ are concatenated
- Allows network to learn noise-dependent behavior
- Simple but effective conditioning mechanism

##### 2. **Batch Normalization**
- Stabilizes training across different noise levels
- Reduces internal covariate shift
- Enables faster convergence

##### 3. **LeakyReLU Activation**
- Prevents dead neurons (negative slope = 0.01)
- Better gradient flow than standard ReLU
- Suitable for score prediction (can be negative)

##### 4. **No Output Activation**
- Score can take any real value
- No constraints on score magnitude or direction
- Linear output allows full expressivity

---

### Forward Pass

```python
def forward(self, x, sigma):
    # Concatenate data and noise level
    x_sigma = torch.cat([x, sigma], dim=1)  # [batch, 3]
    
    # Pass through network
    score = self.net(x_sigma)  # [batch, 2]
    
    return score  # ∇_x log p_σ(x)
```

#### Input-Output Relationship

- **Input**: $(x_\sigma, \sigma)$ where $x_\sigma = x + \epsilon$, $\epsilon \sim \mathcal{N}(0,\sigma^2I)$
- **Output**: $s_\theta(x_\sigma, \sigma) \approx \nabla_x \log p_\sigma(x_\sigma)$
- **Goal**: Predict the direction that points back toward high-density regions

---

### Why This Architecture?

#### MLP vs Other Architectures
- ✅ **Simple**: Easy to implement and debug
- ✅ **Flexible**: Can handle arbitrary noise levels
- ✅ **Fast**: Efficient forward/backward passes
- ✅ **Sufficient**: For 2D data, MLP captures necessary complexity

#### Depth Considerations
- **6 Layers**: Enough capacity to learn complex score functions
- **Not Too Deep**: Avoids overfitting on limited training data
- **Balanced**: Good trade-off between expressivity and trainability

#### Width Considerations
- **256 Neurons**: Sufficient width for smooth score predictions
- **Consistent**: Same width across layers simplifies design

---

### Parameter Count

- **Total Parameters**: ~200K-300K (depending on layer structure)
- **Trainable**: All parameters updated during training
- **Memory**: Efficient for 2D data, scales well

---

### Expected Behavior

#### After Training

1. **Low σ Input**: Should predict sharp, accurate scores pointing to modes
2. **High σ Input**: Should predict smoother scores with broader attraction
3. **Noisy x**: Should predict direction back to clean data region
4. **Out-of-Distribution x**: Should still provide reasonable score estimates

#### Failure Modes

- **Collapse**: Network predicts zeros (no gradient flow)
- **Explosion**: Very large scores (instability, needs gradient clipping)
- **Incorrect Direction**: Scores point away from data (training issue)

## 🏋️ Training Function

### Overview

This cell implements the complete training loop for the score network. It includes essential training mechanisms such as early stopping, learning rate scheduling, gradient clipping, and periodic evaluation through sampling and visualization.

| `patience` | 200 | Early stopping patience |

---

### Training Loop Structure

#### 1. **Initialization**
```python
- Move model to device (CPU/GPU)
- Initialize Adam optimizer
- Setup learning rate scheduler (step decay)
- Initialize tracking variables
```

#### 2. **Epoch Loop**
For each epoch:
- **Batch Loop**: Iterate through data batches
  - Sample noise level σ (fixed or random)
  - Add noise to data: $x_{noisy} = x + \epsilon$
  - Forward pass: predict score $s_\theta(x_{noisy}, \sigma)$
  - Compute DSM loss
  - Backward pass: compute gradients
  - Gradient clipping (if enabled)
  - Update parameters

#### 3. **Learning Rate Scheduling**
- StepLR scheduler: Reduces LR by `scheduler_gamma` every `scheduler_step` epochs
- Example: LR = 0.001 → 0.0005 → 0.00025 ...

#### 4. **Early Stopping**
- Tracks best loss seen so far
- If no improvement for `patience` epochs, stops training
- Loads best model checkpoint at the end

#### 5. **Periodic Evaluation**
- Every 100 epochs (and epoch 1):
  - Prints current loss
  - Performs sampling from current model
  - Visualizes samples vs ground truth
  - Helps monitor training progress

#### 6. **Model Checkpointing**
- Saves best model based on validation loss
- Loads best model at end of training

---

### Training Strategies

#### Fixed Sigma Training (`fixed_sigma != None`)
- **Use Case**: Training models specialized for one noise level
- **Advantage**: Faster convergence, optimized for specific σ
- **Disadvantage**: Cannot generalize to other noise levels
- **Example**: Train separate models for σ = 1, 3, 7

#### Varying Sigma Training (`fixed_sigma == None`)
- **Use Case**: Training a general-purpose score network
- **Advantage**: Learns scores for multiple noise scales simultaneously
- **Disadvantage**: Slower convergence, requires more capacity
- **Noise Sampling**: Random σ ∈ [1, 20] for each batch

---

### Training Dynamics

#### Loss Evolution
- **Initial**: ~0.3-0.5 (random predictions)
- **Middle**: ~0.05-0.1 (learning score directions)

## 🎲 Sampling Methods

### Overview

This cell implements three different sampling strategies that use the learned score function to generate samples from the data distribution. Each method has different trade-offs in terms of sample quality, diversity, and computational cost.

---

### 1. Annealed Langevin Dynamics

#### Function: `annealed_sampling()`

**Concept**: Gradually reduces noise level during sampling, combining exploration (high σ) with exploitation (low σ).

#### Process

1. **Initialization**: Start with samples at random positions
2. **Annealing Schedule**: Progressively reduce σ from `sigma_start` to `sigma_end`
   - Typically: σ = 20 → 19 → 18 → ... → 1
   - `num_sigmas`: Number of noise levels to traverse
3. **At Each Noise Level**:
   - Run `steps_per_sigma` Langevin steps
   - Use current σ for score prediction
   - Gradually refine samples toward data distribution
4. **Final Sample**: Result after annealing complete

#### Algorithm

```python
for σ in [σ_start, σ_mid, ..., σ_end]:
    for step in range(steps_per_sigma):
        score = model(x, σ)
        noise = N(0, √(2ε·σ²))  # Noise scales with σ
        x = x + ε·score + noise  # Langevin step
```

#### Advantages
- ✅ **Best Quality**: Combines broad exploration with fine refinement
- ✅ **Diverse Samples**: High σ allows escaping local modes
- ✅ **Robust**: Works well even with imperfect score estimates

#### Disadvantages
- ❌ **Slow**: Many steps required (num_sigmas × steps_per_sigma)
- ❌ **Parameter Sensitive**: Need to tune annealing schedule

#### Parameters
- `sigma_start`: 20 (high noise for exploration)
- `sigma_end`: 1 (low noise for refinement)
- `num_sigmas`: 20 (number of noise levels)
- `steps_per_sigma`: 50 (steps per noise level)
- `step_size`: 0.1 (gradient step size)

#### Theoretical Basis

**Annealing Principle**: 
- High σ → smooth distribution → easy to explore
- Low σ → sharp distribution → precise sampling
- Transition → gradual refinement

---

### 2. Standard Langevin Dynamics

#### Function: `langevin_sampling()`

**Concept**: Standard Langevin MCMC with fixed noise level.

#### Process

1. **Initialization**: Start from random points
2. **Iteration**: For `num_steps`:
   - Predict score at current position
   - Take gradient step (score direction)
   - Add stochastic noise
   - Update position

#### Algorithm

```python
for step in range(num_steps):
    score = model(x, σ=1)  # Fixed noise level
    noise = N(0, √(2ε))
    x = x + ε·score + noise
```

#### Langevin Dynamics Formula

$$x_{t+1} = x_t + \frac{\epsilon}{2}\nabla_x \log p(x_t) + \sqrt{\epsilon} z_t$$

Where $z_t \sim \mathcal{N}(0, I)$ is the random noise.

#### Advantages
- ✅ **Faster**: Fewer steps than annealing
- ✅ **Simple**: Easy to implement and understand
- ✅ **Stochastic**: Maintains diversity through noise

#### Disadvantages
- ❌ **Mode Stuck**: May get trapped in local modes
- ❌ **Quality**: Lower than annealed sampling
- ❌ **σ Dependent**: Requires correct noise level

#### Parameters
- `num_steps`: 50 (number of Langevin iterations)
- `step_size`: 0.1 (gradient step size)
- `σ`: Fixed at 1 (must match training)

---

### 3. Deterministic Sampling

#### Function: `deterministic_sampling()`

**Concept**: Pure gradient ascent without noise—follows score direction deterministically.

#### Process

1. **Initialization**: Start from random points
2. **Iteration**: For `num_steps`:
   - Predict score at current position
   - Take gradient step (no noise)
   - Update position

#### Algorithm

```python
for step in range(num_steps):
    score = model(x, σ=1)
    x = x + ε·score  # No noise term
```

#### Advantages
- ✅ **Fastest**: No randomness to compute
- ✅ **Deterministic**: Reproducible results
- ✅ **Direct Path**: Follows shortest path to mode

#### Disadvantages
- ❌ **Mode Collapse**: Gets stuck in nearest mode
- ❌ **No Exploration**: Cannot escape local minima
- ❌ **Limited Diversity**: All samples converge to same regions

#### Use Cases
- Fast generation when diversity not critical
- Demonstrating convergence behavior
- Comparing with stochastic methods

---

### Comparison Summary

| Method | Quality | Speed | Diversity | Best For |
|--------|---------|-------|-----------|----------|
| **Annealed** | ⭐⭐⭐⭐⭐ | ⭐⭐ | ⭐⭐⭐⭐⭐ | Best samples |
| **Langevin** | ⭐⭐⭐⭐ | ⭐⭐⭐⭐ | ⭐⭐⭐⭐ | Balanced |
| **Deterministic** | ⭐⭐⭐ | ⭐⭐⭐⭐⭐ | ⭐⭐ | Quick preview |

---

### Implementation Details

#### Gradient Step Size (ε)
- Controls how far to move per step
- Too large: Overshooting, instability
- Too small: Slow convergence
- Typical: 0.1-0.2 for 2D data

#### Noise Scaling
- Annealed: Noise scales with σ (`√(2εσ²)`)
- Langevin: Fixed noise (`√(2ε)`)
- Deterministic: No noise

#### Convergence Criteria
- In practice, run fixed number of steps
- Could add early stopping based on score magnitude
- Visual inspection is common validation method

- Good coverage, may miss some regions
- Stochastic variation in final positions
- Faster but less complete than annealing

#### Deterministic Sampling
- Samples converge to nearest mode

- Clear directional paths visible in trajectories

## 🌐 Phase 3: Varying Sigma Training

### Overview

This section implements **Phase 3**, the most powerful training strategy: training a single score network on **multiple noise levels simultaneously**. This approach learns a generalizable score function that works across different noise scales, providing flexibility and robustness unmatched by fixed-sigma training.

---

### Why Varying Sigma Training?

#### Limitations of Fixed Sigma (Phase 1)

❌ **Specialization**: Each model only works at one noise level
❌ **Inefficiency**: Need separate models for different σ
❌ **Rigidity**: Cannot adapt to different sampling needs
❌ **Storage**: Multiple model checkpoints required

#### Advantages of Varying Sigma (Phase 3)

✅ **Generalization**: One model handles multiple noise scales
✅ **Flexibility**: Same model works for different σ during sampling
✅ **Robustness**: More stable across noise level variations
✅ **Efficiency**: Single model, smaller storage footprint

---

### Training Configuration

#### Extended Training Setup

| Parameter | Value | Rationale |
|-----------|-------|-----------|
| **Epochs** | 1000 | More epochs needed for multi-scale learning |
| **Learning Rate** | 0.001 | Standard Adam rate |
| **Batch Size** | 256 | Efficient batch processing |
| **Scheduler Step** | 500 | LR decay at half and three-quarter points |
