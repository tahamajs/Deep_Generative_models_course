```python



class ResidualConvBlock(nn.Module):
    def __init__(
        self, in_channels: int, out_channels: int, is_res: bool = False
    ) -> None:
        super().__init__()

        self.same_channels = in_channels == out_channels

        self.is_res = is_res

        self.conv1 = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, 1, 1),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
        )

        self.conv2 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, 3, 1, 1),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:

        if self.is_res:
            x1 = self.conv1(x)

            x2 = self.conv2(x1)

            if self.same_channels:
                out = x + x2
            else:
                shortcut = nn.Conv2d(x.shape[1], x2.shape[1], kernel_size=1, stride=1, padding=0).to(x.device)
                out = shortcut(x) + x2

            return out / 1.414

        else:
            x1 = self.conv1(x)
            x2 = self.conv2(x1)
            return x2

    def get_out_channels(self):
        return self.conv2[0].out_channels

    def set_out_channels(self, out_channels):
        self.conv1[0].out_channels = out_channels
        self.conv2[0].in_channels = out_channels
        self.conv2[0].out_channels = out_channels
```

## Residual Convolutional Block

This cell defines a residual convolutional block used in the U-Net architecture. It includes two convolutional layers with batch normalization and GELU activation, with an optional residual connection for improved gradient flow. The block supports both residual and non-residual modes.

```python
class UnetUp(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(UnetUp, self).__init__()

        layers = [
            nn.ConvTranspose2d(in_channels, out_channels, 2, 2),
            ResidualConvBlock(out_channels, out_channels),
            ResidualConvBlock(out_channels, out_channels),
        ]

        self.model = nn.Sequential(*layers)

    def forward(self, x, skip):
        x = torch.cat((x, skip), 1)

        x = self.model(x)
        return x
```

## U-Net Up Block

This cell defines the upsampling block for the U-Net decoder. It performs transposed convolution followed by two residual convolutional blocks to upsample and refine features, concatenating skip connections from the encoder.

```python

class UnetDown(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(UnetDown, self).__init__()

        layers = [ResidualConvBlock(in_channels, out_channels), ResidualConvBlock(out_channels, out_channels), nn.MaxPool2d(2)]

        self.model = nn.Sequential(*layers)

    def forward(self, x):
        return self.model(x)
```

## U-Net Down Block

This cell defines the downsampling block for the U-Net encoder. It applies two residual convolutional blocks followed by max pooling to reduce spatial dimensions and extract hierarchical features.

```python

class EmbedFC(nn.Module):
    def __init__(self, input_dim, emb_dim):
        super(EmbedFC, self).__init__()

        self.input_dim = input_dim

        layers = [
            nn.Linear(input_dim, emb_dim),
            nn.GELU(),
            nn.Linear(emb_dim, emb_dim),
        ]

        self.model = nn.Sequential(*layers)

    def forward(self, x):
        x = x.view(-1, self.input_dim)
        return self.model(x)
```

## Embedding Fully Connected Layer

This cell defines an embedding layer for time and context conditioning. It uses linear layers with GELU activation to embed scalar time steps or context vectors into higher-dimensional spaces for conditioning the U-Net.

```python



class Unet(nn.Module):
    def __init__(self, in_channels, n_feat=64, n_cfeat=5, height=16):
        super(Unet, self).__init__()

        self.in_channels = in_channels
        self.n_feat = n_feat
        self.n_cfeat = n_cfeat
        self.h = height 

        self.init_conv = ResidualConvBlock(in_channels, n_feat, is_res=True)

        self.down1 = UnetDown(n_feat, n_feat)
        self.down2 = UnetDown(n_feat, 2 * n_feat)

        self.to_vec = nn.Sequential(nn.AvgPool2d((4)), nn.GELU())

        self.timeembed1 = EmbedFC(1, 2*n_feat)
        self.timeembed2 = EmbedFC(1, 1*n_feat)
        self.contextembed1 = EmbedFC(n_cfeat, 2*n_feat)
        self.contextembed2 = EmbedFC(n_cfeat, 1*n_feat)

        self.up0 = nn.Sequential(
            nn.ConvTranspose2d(2 * n_feat, 2 * n_feat, self.h//4, self.h//4),
            nn.GroupNorm(8, 2 * n_feat), 
            nn.ReLU(),
        )
        self.up1 = UnetUp(4 * n_feat, n_feat)
        self.up2 = UnetUp(2 * n_feat, n_feat)

        self.out = nn.Sequential(
            nn.Conv2d(2 * n_feat, n_feat, 3, 1, 1),
            nn.GroupNorm(8, n_feat),
            nn.ReLU(),
            nn.Conv2d(n_feat, self.in_channels, 3, 1, 1),
        )

    def forward(self, x, t, c=None):

        x = self.init_conv(x)
        down1 = self.down1(x)       
        down2 = self.down2(down1)  

        hiddenvec = self.to_vec(down2)

        if c is None:
            c = torch.zeros(x.shape[0], self.n_cfeat).to(x)

        t = t.unsqueeze(1)  
        cemb1 = self.contextembed1(c).view(-1, self.n_feat * 2, 1, 1)
        temb1 = self.timeembed1(t).view(-1, self.n_feat * 2, 1, 1)
        cemb2 = self.contextembed2(c).view(-1, self.n_feat, 1, 1)
        temb2 = self.timeembed2(t).view(-1, self.n_feat, 1, 1)

        up1 = self.up0(hiddenvec)
        up2 = self.up1(cemb1 * up1 + temb1, down2)
        up3 = self.up2(cemb2 * up2 + temb2, down1)
        out = self.out(torch.cat((up3, x), 1))
        return out
```

## Understanding the U-Net Architecture

The U-Net is the core denoising network in diffusion models. Here's a conceptual overview:

### Architecture Components:

1. **Encoder (Downsampling Path)**: 
   - Progressively reduces spatial dimensions
   - Extracts hierarchical features
   - Uses residual blocks and max pooling

2. **Bottleneck**:
   - Processes the most abstract representation
   - Incorporates time and context embeddings

3. **Decoder (Upsampling Path)**:
   - Reconstructs the image
   - Uses skip connections from encoder
   - Gradually increases spatial resolution

4. **Time Embedding**:
   - Conditions the network on diffusion timestep
   - Allows the network to adapt denoising strength

5. **Context Embedding**:
   - Enables conditional generation (e.g., class labels)
   - Provides class information for controlled generation

```python
@torch.no_grad()
def sample_ddpm(
    nn_model, n_sample, timesteps, beta, alpha, alpha_bar, device, height=16
):
    nn_model.eval()
    samples = torch.randn(n_sample, 3, height, height).to(device)

    for t in tqdm(reversed(range(timesteps)), desc="DDPM Sampling"):
        t_batch = torch.full((n_sample,), t, dtype=torch.long, device=device)

        t_normalized = t_batch.float() / timesteps
        noise_pred = nn_model(samples, t_normalized)

        sqrt_alpha_t = torch.sqrt(alpha[t]).reshape(1, 1, 1, 1)
        sqrt_one_minus_alpha_bar_t = torch.sqrt(1 - alpha_bar[t]).reshape(1, 1, 1, 1)
        sqrt_recip_alpha_t = torch.sqrt(1.0 / alpha[t]).reshape(1, 1, 1, 1)
        beta_t = beta[t]
        sqrt_beta_t = torch.sqrt(beta[t]).reshape(1, 1, 1, 1)

        mean = sqrt_recip_alpha_t * (
            samples - (beta_t / sqrt_one_minus_alpha_bar_t) * noise_pred
        )

        if t > 0:
            noise = torch.randn_like(samples).to(device)
            sigma_t = sqrt_beta_t
            samples = mean + sigma_t * noise
        else:
            samples = mean

    return samples
```

## DDPM Sampling Function

This cell implements the reverse diffusion sampling for Denoising Diffusion Probabilistic Models (DDPM). It starts from pure noise and iteratively denoises using the trained model to generate new images, following the original DDPM algorithm.

```python
@torch.no_grad()
def sample_ddim(
    nn_model,
    n_sample,
    timesteps,
    beta,
    alpha,
    alpha_bar,
    device,
    height=16,
    eta=0.0,
    n_steps=50,
):
    nn_model.eval()
    samples = torch.randn(n_sample, 3, height, height).to(device)

    n = n_steps
    step_size = timesteps // n

    print(f"Starting DDIM sampling with {n} steps (step size: {step_size})")

    for i in tqdm(reversed(range(n)), desc="DDIM Sampling"):
        t = i * step_size
        t_prev = (i - 1) * step_size if i > 0 else 0

        t_batch = torch.full((n_sample,), t, dtype=torch.long, device=device)
        t_prev_batch = torch.full((n_sample,), t_prev, dtype=torch.long, device=device)

        t_normalized = t_batch.float() / timesteps
        noise_pred = nn_model(samples, t_normalized)

        alpha_t = alpha[t].reshape(1, 1, 1, 1)
        alpha_bar_t = alpha_bar[t].reshape(1, 1, 1, 1)
        alpha_bar_prev = alpha_bar[t_prev].reshape(1, 1, 1, 1)

        sqrt_alpha_bar_prev = torch.sqrt(alpha_bar_prev)
        sqrt_alpha_bar_t = torch.sqrt(alpha_bar_t)
        ratio = sqrt_alpha_bar_prev / sqrt_alpha_bar_t

        epsilon = 1e-5
        denom = torch.sqrt(1 - alpha_bar_t)
        denom = torch.where(
            denom < epsilon, torch.tensor(epsilon, device=device), denom
        )

        mean = ratio * samples - (beta[t].reshape(1, 1, 1, 1) / denom) * noise_pred

        if eta > 0.0 and t > 0:
            sigma_t = eta * torch.sqrt(beta[t]).reshape(1, 1, 1, 1)
            noise = torch.randn_like(samples).to(device)
            samples = mean + sigma_t * noise
            print(f"Step {i+1}/{n}: Added noise with sigma_t={sigma_t.item()}")
        else:
            samples = mean
            print(f"Step {i+1}/{n}: Deterministic step")

    print("DDIM sampling completed.")
    return samples
```

This visualization demonstrates how a clean image progressively becomes pure noise through the forward diffusion process. We'll show an example image from the dataset at different noise levels (timesteps).


```python
def visualize_forward_diffusion(image, alpha_bar, timesteps=1000, num_steps=10):
    """Visualize how an image gets progressively noisier during forward diffusion"""
    image = image.to(device)
    selected_timesteps = [int(timesteps * i / (num_steps - 1)) for i in range(num_steps)]
    selected_timesteps[-1] = timesteps - 1  # Ensure we include the last timestep
    
    fig, axes = plt.subplots(2, 5, figsize=(20, 8))
    axes = axes.flatten()
    
    for idx, t in enumerate(selected_timesteps):
        # Apply forward diffusion
        noise = torch.randn_like(image).to(device)
        sqrt_alpha_bar_t = torch.sqrt(alpha_bar[t]).item()
        sqrt_one_minus_alpha_bar_t = torch.sqrt(1 - alpha_bar[t]).item()
        noisy_image = sqrt_alpha_bar_t * image + sqrt_one_minus_alpha_bar_t * noise
        
        # Normalize for display
        img_display = unorm(noisy_image[0].cpu()).permute(1, 2, 0).numpy()
        axes[idx].imshow(img_display, interpolation='nearest')
        axes[idx].set_title(f't = {t}\nᾱ = {alpha_bar[t].item():.4f}', fontsize=10)
        axes[idx].axis('off')
    
    plt.suptitle('Forward Diffusion: Progressive Noise Addition', fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()

# Visualize forward diffusion if data is available
if 'train_dataset' in globals() and len(train_dataset) > 0:
    sample_image, _ = train_dataset[0]
    sample_image = sample_image.unsqueeze(0).to(device)
    
    if 'alpha_bar' in globals():
        print("Visualizing forward diffusion process on a sample image...")
        visualize_forward_diffusion(sample_image, alpha_bar, timesteps, num_steps=10)
    else:
        print("Note: alpha_bar needs to be computed first (run diffusion schedule cell)")
else:
    print("Note: Dataset needs to be loaded first to visualize forward diffusion")
```

## Visualization: Reverse Diffusion (Denoising Process)

After training, we can visualize how the model denoises an image step-by-step during the reverse diffusion process. This shows the generation process from pure noise to a clean image.


```python
@torch.no_grad()
def visualize_reverse_diffusion(nn_model, n_sample=1, timesteps=1000, beta=None, alpha=None, 
                                alpha_bar=None, device=None, height=16, num_steps=10):
    """Visualize the reverse diffusion (denoising) process step by step"""
    if nn_model is None or beta is None:
        print("Note: Model and diffusion schedule need to be available")
        return
    
    nn_model.eval()
    samples = torch.randn(n_sample, 3, height, height).to(device)
    
    selected_steps = [int(i * timesteps / (num_steps - 1)) for i in range(num_steps)]
    selected_steps[-1] = timesteps - 1
    
    denoising_steps = []
    step_indices = []
    
    for i, t in enumerate(reversed(range(timesteps))):
        t_batch = torch.full((n_sample,), t, dtype=torch.long, device=device)
        t_normalized = t_batch.float() / timesteps
        noise_pred = nn_model(samples, t_normalized)
        
        # DDPM sampling step
        sqrt_alpha_t = torch.sqrt(alpha[t]).reshape(1, 1, 1, 1)
        sqrt_one_minus_alpha_bar_t = torch.sqrt(1 - alpha_bar[t]).reshape(1, 1, 1, 1)
        sqrt_recip_alpha_t = torch.sqrt(1.0 / alpha[t]).reshape(1, 1, 1, 1)
        beta_t = beta[t]
        
        mean = sqrt_recip_alpha_t * (
            samples - (beta_t / sqrt_one_minus_alpha_bar_t) * noise_pred
        )
        
        if t > 0:
            noise = torch.randn_like(samples).to(device)
            sqrt_beta_t = torch.sqrt(beta[t]).reshape(1, 1, 1, 1)
            sigma_t = sqrt_beta_t
            samples = mean + sigma_t * noise
        else:
            samples = mean
        
        # Store intermediate steps
        if t in selected_steps:
            denoising_steps.append(samples.clone().cpu())
            step_indices.append(timesteps - 1 - i)
    
    # Visualize
    fig, axes = plt.subplots(2, 5, figsize=(20, 8))
    axes = axes.flatten()
    
    for idx, (img, step) in enumerate(zip(denoising_steps, step_indices)):
        img_display = unorm(img[0]).permute(1, 2, 0).numpy()
        axes[idx].imshow(img_display, interpolation='nearest')
        axes[idx].set_title(f'Step {step}\nt = {timesteps - 1 - step}', fontsize=10)
        axes[idx].axis('off')
    
    plt.suptitle('Reverse Diffusion: Progressive Denoising', fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()
    
    return denoising_steps

print("Reverse diffusion visualization function defined. Call this after model training.")
```

## DDPM vs DDIM: Key Differences

### DDPM (Denoising Diffusion Probabilistic Models)
- **Sampling**: Stochastic (uses random noise at each step)
- **Steps**: Requires all 1000 timesteps for full quality
- **Speed**: Slower but produces high-quality samples
- **Use Case**: Best for final generation when quality is paramount

### DDIM (Denoising Diffusion Implicit Models)
- **Sampling**: Deterministic or semi-stochastic (controlled by η parameter)
- **Steps**: Can use fewer steps (e.g., 50-500) with acceptable quality
- **Speed**: Faster generation through step skipping
- **Use Case**: Useful for fast prototyping and interactive applications

### Key Insight
Both methods use the same trained model but differ in sampling strategy:
- **DDPM**: Follows the learned reverse diffusion process exactly
- **DDIM**: Uses a deterministic trajectory that can skip steps

## Mathematical Foundations

Understanding the key equations helps clarify how diffusion models work:

### Forward Diffusion
$$q(x_t | x_{t-1}) = \mathcal{N}(x_t; \sqrt{1-\beta_t} x_{t-1}, \beta_t I)$$

This adds Gaussian noise at each step. After $T$ steps:
$$q(x_T | x_0) = \mathcal{N}(x_T; \sqrt{\bar{\alpha}_T} x_0, (1-\bar{\alpha}_T) I)$$

where $\bar{\alpha}_t = \prod_{s=1}^t (1-\beta_s)$ is the cumulative product.

### Reverse Diffusion (What We Learn)
$$p_\theta(x_{t-1} | x_t) = \mathcal{N}(x_{t-1}; \mu_\theta(x_t, t), \Sigma_\theta(x_t, t))$$

The model learns to predict:
- **Noise prediction**: $\epsilon_\theta(x_t, t)$ - the noise added to $x_0$
- **Mean prediction**: Used to reconstruct $x_{t-1}$ from $x_t$

### Training Objective
$$\mathcal{L} = \mathbb{E}_{t,x_0,\epsilon} \left[ ||\epsilon - \epsilon_\theta(x_t, t)||^2 \right]$$

We train the U-Net to predict the noise $\epsilon$ that was added to get $x_t$ from $x_0$.

### DDPM Sampling
$$x_{t-1} = \frac{1}{\sqrt{\alpha_t}} \left( x_t - \frac{\beta_t}{\sqrt{1-\bar{\alpha}_t}} \epsilon_\theta(x_t, t) \right) + \sqrt{\beta_t} z$$

where $z \sim \mathcal{N}(0, I)$ is random noise for stochasticity.

### DDIM Sampling
$$x_{t-1} = \sqrt{\bar{\alpha}_{t-1}} \left( \frac{x_t - \sqrt{1-\bar{\alpha}_t} \epsilon_\theta(x_t, t)}{\sqrt{\bar{\alpha}_t}} \right) + \sqrt{1-\bar{\alpha}_{t-1} - \sigma_t^2} \epsilon_\theta(x_t, t)$$

DDIM uses a deterministic trajectory when $\eta=0$ (no noise term).

## Understanding Diffusion Model Concepts - Visual Guide

### The Core Idea

Diffusion models learn to reverse a gradual corruption process:
1. **Forward Process**: Clean image → Noise (we know how to do this)
2. **Reverse Process**: Noise → Clean image (this is what we learn!)

### Why This Works

- **Simple Forward Process**: Easy to add noise at any timestep
- **Rich Representations**: U-Net learns powerful image features during denoising
- **Stable Training**: Learning to denoise is more stable than other generative models

### Key Innovations

1. **Noise Schedule**: Carefully designed to preserve information while adding noise
2. **Time Conditioning**: Model adapts denoising strength based on timestep
3. **Iterative Refinement**: Each step improves the image slightly
4. **Flexible Sampling**: DDIM allows faster generation with fewer steps

### Applications

- **Image Generation**: Create new images from scratch
- **Image Editing**: Modify existing images by controlling the diffusion process
- **Super-Resolution**: Enhance image quality through iterative refinement
- **Inpainting**: Fill in missing parts of images
